```python
# INIZIO PROGETTO

import pandas as pd
df = pd.read_csv("wiki_movie_plots_deduped.csv")
df = df[['Genre', 'Plot']]
```

```python
import re

# definisco una funzione che pulisca i testi
def pulisci_testo(text):
    text = text.lower()                      # minuscolo
    text = re.sub(r'\d+', '', text)          # rimuove numeri
    text = re.sub(r'[^\w\s]', '', text)      # rimuove punteggiatura
    text = text.strip()                      # rimuove spazi ai bordi
    return text

```

```python
# applico la funzione a 5 testi e vedo il confronto originale/pulito
for i in range(1000, 1005):
    original = df.loc[i, 'Plot']
    cleaned = pulisci_testo(original)
    print(f"Plot {i}:\n")            # f sta per formatted string
    print("Originale:", original)
    print()
    print("Pulito:", cleaned)
    print()  # Riga vuota
```

```python
# Crea una colonna nel dataset con i plot puliti
df['Cleaned_Plot'] = df['Plot'].apply(pulisci_testo)
df
```

```python
# Lemmatizzazione
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet', quiet=True)

# Inizializza il lemmatizzatore
lemmatizer = WordNetLemmatizer()

def lemmatize_text(text):
    # Tokenizzazione semplice (divide su spazi)
    tokens = text.split()
    
    # Lemmatizza ogni token
    lemmatized_tokens = [lemmatizer.lemmatize(token) for token in tokens]
    
    # Riunisce i token in una stringa
    return ' '.join(lemmatized_tokens)

df['Lemmatized_Plot'] = df['Cleaned_Plot'].apply(lemmatize_text)

# Visione confronto testo pulito e lemmatizzato
for i in range(3):
    idx = df.index[i]
    print(f"\nPlot {idx}:")
    print("Testo pulito:\n", df.loc[idx, 'Cleaned_Plot'][:100], "...")
    print("\nTesto lemmatizzato:\n", df.loc[idx, 'Lemmatized_Plot'][:100], "...")
```
