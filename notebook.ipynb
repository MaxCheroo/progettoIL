```python
# INIZIO PROGETTO
#importazione csv
import pandas as pd
df = pd.read_csv("wiki_movie_plots_deduped.csv")
df
```

```python
# filtro colonne
df = df[['Genre', 'Plot']]
df
```

```python
# print(df['Plot'].head())
# leggo uno specifico plot
plot1 = df.loc[1, 'Plot']
plot1
```

```python
# Guardo tutti i generi e il loro numero convertendo in dizionario
allgenres = df['Genre'].value_counts()
allgenres_dict = allgenres.to_dict()
allgenres_dict
```

```python
# Filtro il dataset tenendo solo 4 generi
selected_genres = ['drama', 'comedy', 'horror', 'action']
df = df[df['Genre'].isin(selected_genres)].copy()
print(df['Genre'].value_counts())
df
```

```python
df_bilanciato = (
    df.groupby('Genre')
    .apply(lambda x: x.sample(n=1100, random_state=42) if len(x) > 1100 else x)
    .reset_index(drop=True)
)

# mescola le righe
df_bilanciato = df_bilanciato.sample(frac=1, random_state=42).reset_index(drop=True)


#  Visualizza un riepilogo
print(df_bilanciato['Genre'].value_counts())
df_bilanciato
```

```python
import re

# definisco una funzione che pulisca i testi
def pulisci_testo(text):
    text = text.lower()                      # minuscolo
    text = re.sub(r'\d+', '', text)          # rimuove numeri
    text = re.sub(r'[^\w\s]', '', text)      # rimuove punteggiatura
    text = text.strip()                      # rimuove spazi ai bordi
    return text
```

```python
# applico la funzione a 3 testi e vedo il confronto originale/pulito
for i in range(1, 4):
    original = df_bilanciato.loc[i, 'Plot']
    cleaned = pulisci_testo(original)
    print(f"Plot {i}:\n")            # f sta per formatted string
    print("Originale:", original)
    print("Pulito:", cleaned)
    print()  # Riga vuota
```
