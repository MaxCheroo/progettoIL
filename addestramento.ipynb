```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
import joblib

# Dataset già pulito nel preprocessing
df = pd.read_csv("datasetpronto.csv")
testi = df['Cleaned_Plot'].tolist()
etichette = df['Genre'].tolist()

vectorizer = TfidfVectorizer(stop_words='english')
X = vectorizer.fit_transform(testi)

model = SVC(kernel='linear', C=1, probability=True)
model.fit(X, etichette)

joblib.dump(model, "modello_svc.joblib")
joblib.dump(vectorizer, "vectorizer_tfidf.joblib")
```
